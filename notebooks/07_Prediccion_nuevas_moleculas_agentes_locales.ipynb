Este notebook muestra cómo entraría un set pequeño de moléculas al modelo de tesis usando el panel local sin API.

Requisito: las moléculas deben venir con los 11 descriptores alvaDesc/WEKA seleccionados en la tesis.

## Qué debe cambiar el usuario

Para evaluar moléculas nuevas, cambia la ruta `data/query/example_positive_negative_11.csv` por tu archivo. El archivo debe tener los 11 descriptores de tesis. Si además tiene SMILES, el `ChemicalAgent` podrá validar estructura y Lipinski; si no tiene SMILES, la predicción igual funciona con los descriptores.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
%cd $ROOT
```

## Ejemplo mínimo

Se usan dos moléculas de ejemplo tomadas del conjunto externo: una positiva (`Act1`) y una negativa (`Act-1`).

```python
example_path = Path('data/query/example_positive_negative_11.csv')
if not example_path.exists():
    external = pd.read_csv('data/raw/external_11.csv')
    example = pd.concat([
        external[external['Actividad'].eq('Act1')].head(1),
        external[external['Actividad'].eq('Act-1')].head(1),
    ], ignore_index=True)
    example.insert(0, 'example_id', ['positive_external_example', 'negative_external_example'])
    example_path.parent.mkdir(parents=True, exist_ok=True)
    example.to_csv(example_path, index=False)
pd.read_csv(example_path)
```
