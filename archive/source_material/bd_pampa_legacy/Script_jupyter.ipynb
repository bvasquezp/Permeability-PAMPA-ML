```python
#  Paso 1: Importar librerías
# =============================================================================
import pandas as pd
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
import numpy as np
```

```python
#  Paso 2: Cargar datos y definir características seleccionadas
# =============================================================================

# --- Configuración Inicial ---
# Ruta a tu archivo CSV de entrenamiento de fingerprints.
ruta_archivo_csv = r"C:\Users\benja\Desktop\BD PAMPA\Calculos Fingerprint\Bases de datos\new_training.csv"

# Nombre de la columna de la clase/actividad.
columna_clase = "Actividad"

# Lista de los 20 atributos seleccionados por J48 + BestFirst.
atributos_seleccionados = [
    'MACCSFP8', 'MACCSFP25', 'MACCSFP34', 'MACCSFP40', 'MACCSFP43',
    'MACCSFP47', 'MACCSFP49', 'MACCSFP50', 'MACCSFP54', 'MACCSFP56',
    'MACCSFP62', 'MACCSFP70', 'MACCSFP76', 'MACCSFP80', 'MACCSFP84',
    'MACCSFP99', 'MACCSFP112', 'MACCSFP125', 'MACCSFP156', 'MACCSFP161'
]

print(f"Cargando datos desde: {ruta_archivo_csv}")

# --- Carga y Preparación de Datos ---
df_completo = pd.read_csv(ruta_archivo_csv)
X = df_completo[atributos_seleccionados]
y = df_completo[columna_clase]

print(f"Se usarán {X.shape[1]} atributos seleccionados para {X.shape[0]} moléculas.")
```

```python
#Paso 3: Definir los modelos a evaluar
# =============================================================================
modelos = {
    "Árbol de Decisión (J48)": DecisionTreeClassifier(random_state=42),
    "Regresión Logística": LogisticRegression(max_iter=1000, random_state=42),
    "k-NN (IBk, k=5)": KNeighborsClassifier(n_neighbors=5),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC(probability=True, random_state=42)
}
```

```python
#  Paso 4: Entrenar y Evaluar con Validación Cruzada (10-folds)
# =============================================================================
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
resultados = {}

print("\n--- Iniciando Evaluación de Modelos ---")

for nombre, modelo in modelos.items():
    scores_accuracy = cross_val_score(modelo, X, y, cv=cv, scoring='accuracy')
    scores_roc_auc = cross_val_score(modelo, X, y, cv=cv, scoring='roc_auc')
    
    resultados[nombre] = {
        'Accuracy': np.mean(scores_accuracy),
        'ROC AUC': np.mean(scores_roc_auc)
    }
    print(f"  > Evaluado: {nombre}")

print("--- Evaluación Completada ---\n")
```

```python
#  Paso 5: Mostrar resultados
# =============================================================================
df_resultados = pd.DataFrame.from_dict(resultados, orient='index')

print("📊 Tabla Comparativa de Rendimiento (Fingerprints):")
print(df_resultados.round(3))
```

```python
#PASO 6: GUARDAR LOS RESULTADOS EN UN ARCHIVO (NUEVO)
# =============================================================================
nombre_archivo_salida = 'resultados_fingerprints_training.csv'
df_resultados.to_csv(nombre_archivo_salida)

print(f"\n✅ ¡Tabla de resultados guardada exitosamente en el archivo '{nombre_archivo_salida}'!")
```
