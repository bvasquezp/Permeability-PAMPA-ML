```python
import pandas as pd
import sys

# =============================================================================
#  Configuración de Rutas
# =============================================================================
# Ruta al archivo de DMs 
ruta_dm_csv = r"C:\Users\benja\Desktop\BD PAMPA\Calculos descriptores moleculares\bases de dato\test_dm.csv"

# Ruta al archivo de Fingerprints 
ruta_fp_csv = r"C:\Users\benja\Desktop\BD PAMPA\Calculos Fingerprint\Bases de datos\new_test.csv" 

nombre_salida_csv = "fusion_test_set.csv"

# =============================================================================
#  Código de Fusión
# =============================================================================
try:
    print("--- PASO 1: Cargando Descriptores Moleculares ---")
    # Instrucción explícita para leer usando coma (,)
    df_dm = pd.read_csv(ruta_dm_csv, sep=',')
    print(f"  > Archivo DM cargado. Dimensiones: {df_dm.shape[0]} filas, {df_dm.shape[1]} columnas.")

    print("\n--- PASO 2: Cargando Fingerprints ---")
    df_fp = pd.read_csv(ruta_fp_csv, sep=',')
    print(f"  > Archivo FP cargado. Dimensiones: {df_fp.shape[0]} filas, {df_fp.shape[1]} columnas.")

    # Verificación de consistencia
    if len(df_dm) != len(df_fp):
        print("\n❌ ¡ERROR CRÍTICO! El número de filas no coincide.")
        sys.exit()

    print("\n--- PASO 3: Combinando los datasets ---")
    # Eliminar la columna de 'Actividad' de los DMs para evitar duplicados
    df_dm_features = df_dm.drop(columns=['Actividad'], errors='ignore')
    
    # Unir las características de DMs con el dataframe completo de FPs
    # Reseteamos el índice en ambos para asegurar una unión limpia por posición
    df_combined = pd.concat([df_dm_features.reset_index(drop=True), df_fp.reset_index(drop=True)], axis=1)
    
    print(f"  > Datasets combinados. Dimensiones finales: {df_combined.shape[0]} filas, {df_combined.shape[1]} columnas.")
    
    # Guardar el archivo final usando comas, que es el estándar
    df_combined.to_csv(nombre_salida_csv, index=False, sep=',')
    
    print(f"\n✅ ¡Éxito! Archivo '{nombre_salida_csv}' guardado correctamente usando comas (,) como separador.")

except FileNotFoundError as e:
    print(f"❌ ERROR: No se pudo encontrar el archivo: {e.filename}")
except Exception as e:
    print(f"❌ Ocurrió un error inesperado: {e}")
```
