```python
#  Paso 1: Importar librerías
# =============================================================================
import pandas as pd
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
import numpy as np
```

```python
# --- Definición de variables ---
ruta_archivo_csv = r"C:\Users\benja\Desktop\BD PAMPA\Calculos Fingerprint\Bases de datos\new_training.csv"
columna_clase = "Actividad"
atributos_seleccionados = ['MACCSFP8', 'MACCSFP18', 'MACCSFP25', 'MACCSFP34', 'MACCSFP40', 'MACCSFP43', 'MACCSFP47', 'MACCSFP49', 'MACCSFP50', 'MACCSFP54', 'MACCSFP56', 'MACCSFP62', 'MACCSFP70', 'MACCSFP76', 'MACCSFP80', 'MACCSFP84', 'MACCSFP99', 'MACCSFP112', 'MACCSFP125', 'MACCSFP145', 'MACCSFP156', 'MACCSFP161']

print("CELDA 2: Variables definidas correctamente.")
print(f"La lista de atributos tiene {len(atributos_seleccionados)} elementos.")
```

```python
# --- Carga y Preparación de Datos ---
try:
    df_completo = pd.read_csv(ruta_archivo_csv)
    df_completo.columns = df_completo.columns.str.strip()
    X = df_completo[atributos_seleccionados]
    y = df_completo[columna_clase]
    print(f"CELDA 3: Datos cargados. Se usarán {X.shape[1]} atributos para {X.shape[0]} moléculas.")

except Exception as e:
    print(f"CELDA 3: Ocurrió un error al cargar los datos: {e}")
```

```python
# =============================================================================
#  Paso 3: Definir y Evaluar Modelos
# =============================================================================
modelos = {
    "Árbol de Decisión (J48)": DecisionTreeClassifier(random_state=42),
    "Regresión Logística": LogisticRegression(max_iter=1000, random_state=42),
    "k-NN (IBk, k=5)": KNeighborsClassifier(n_neighbors=5),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC(probability=True, random_state=42)
}
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
resultados = {}

print("\n--- Iniciando Evaluación de Modelos (Fingerprints por Consenso) ---")
for nombre, modelo in modelos.items():
    scores_accuracy = cross_val_score(modelo, X, y, cv=cv, scoring='accuracy')
    scores_roc_auc = cross_val_score(modelo, X, y, cv=cv, scoring='roc_auc')
    
    resultados[nombre] = {
        'Accuracy': np.mean(scores_accuracy),
        'ROC AUC': np.mean(scores_roc_auc)
    }
    print(f"  > Evaluado: {nombre}")
print("--- Evaluación Completada ---\n")

# =============================================================================
#  Paso 4: Mostrar y Guardar resultados
# =============================================================================
df_resultados = pd.DataFrame.from_dict(resultados, orient='index')
print("📊 Tabla Comparativa de Rendimiento (Fingerprints por Consenso):")
print(df_resultados.round(3))

nombre_archivo_salida = 'resultados_modelos_fingerprints_consenso.csv'
df_resultados.to_csv(nombre_archivo_salida)

print(f"\n✅ ¡Tabla de resultados guardada exitosamente en el archivo '{nombre_archivo_salida}'!")
```
